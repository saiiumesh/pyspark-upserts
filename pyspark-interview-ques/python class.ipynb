```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

_**_creating class_**_

```python
class dataValidation:
  def __init__(self,df):
      self.df=df

  def dedup(self,df,keyCol,cdcCol):
      df = df.withColumn("dedup",row_number().over(Window.partitionBy(keyCol).orderBy(desc(cdcCol))))
      df = df.filter(col("dedup") == 1).drop("dedup")
      return df
  
  def dropnulls(self,nullcol):
      df = self.df.filter(col(nullcol).isNotNull())
      return df
      
```

```python
data = [("Alice", 34), ("Bob", 45), ("Cathy", 29)]
schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True)
])
data = [("Alice", 34, 1, "2025-08-01"), ("Bob", 45, 2, "2025-08-02"), ("Cathy", 29, 3, "2025-08-03")]
schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("order_id", IntegerType(), True),
    StructField("order_date", StringType(), True)
])
data = [("Alice", 34, 1, "2025-08-01"), ("Bob", 45, 2, "2025-08-02"), ("Cathy", 29, 3, "2025-08-03"), ("Alice", 34, 4, "2025-08-04"), (None, 50, 5, "2025-08-05"), ("Eve", None, 6, "2025-08-06")]

df = spark.createDataFrame(data, schema)
display(df)
```

```python
cls_obj = dataValidation(df)
```

```python
df = cls_obj.dedup(df,'name','order_id')
df = cls_obj.dropnulls('age')
display(df)
```
